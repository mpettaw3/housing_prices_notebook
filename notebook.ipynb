```python
import pandas as pd
TRAIN_DATASET ="C:\\Users\\Owner\\Downloads\\housing\\train.csv"
TEST_DATASET ="C:\\Users\\Owner\\Downloads\\housing\\test.csv"
training_data = pd.read_csv(TRAIN_DATASET)
test_data = pd.read_csv(TEST_DATASET)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
cat_attribs=["MSZoning",
"Street",
"Alley",
"LotShape",
"LandContour",
"Utilities",
"LotConfig",
"LandSlope",
"Neighborhood",
"Condition1",
"Condition2",
"BldgType",
"HouseStyle",
"RoofStyle",
"RoofMatl",
"Exterior1st",
"Exterior2nd",
"MasVnrType",
"ExterQual",
"ExterCond",
"Foundation",
"BsmtQual",
"BsmtCond",
"BsmtExposure",
"BsmtFinType1",
"BsmtFinType2",
"Heating",
"HeatingQC",
"CentralAir",
"Electrical",
"KitchenQual",
"Functional",
"FireplaceQu",
"GarageType",
"GarageFinish",
"GarageQual",
"GarageCond",
"PavedDrive",
"PoolQC",
"Fence",
"MiscFeature",
"SaleType",
"SaleCondition"]
num_attribs = training_data.columns.drop(cat_attribs).drop(["SalePrice"])
train_labels = training_data["SalePrice"]
dropped_train_salesprice= training_data.drop(columns=["SalePrice"])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ("ord", OrdinalEncoder()),
])

columntrans=ColumnTransformer(
    transformers=[
        ("num", SimpleImputer(strategy="most_frequent"), num_attribs),
        ("cat", cat_pipeline, cat_attribs)
    ]

)
preprocessor = columntrans.fit_transform(dropped_train_salesprice)
```
